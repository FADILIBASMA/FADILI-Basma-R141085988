### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (régression linéaire paramétrique).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives quantitatives $X \in \mathbb{R}^{n \times p}$ (les 10 indicateurs cliniques du jeu *Diabetes*). En sortie, une variable cible continue $\hat{y} \in \mathbb{R}$ mesurant la progression de la maladie.
* **Hypothèse clé :** La relation entre les variables explicatives et la cible est linéaire, et les erreurs (résidus) sont indépendantes, centrées autour de zéro et de variance constante (homoscédasticité).
* **Étapes successives de calcul :** (1) Initialisation du vecteur des paramètres, (2) calcul des prédictions linéaires, (3) minimisation de la somme des carrés des résidus (méthode des moindres carrés ordinaires — MCO) via résolution analytique ou descente de gradient, (4) application des poids optimaux sur le jeu de test.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients de régression $\beta_j$ et l'ordonnée à l'origine $\beta_0$, estimés automatiquement pendant l'entraînement. L'algorithme OLS standard n'a pas d'hyperparamètre de régularisation, mais le choix d'inclure ou non l'ordonnée à l'origine (`fit_intercept=True/False`) constitue un **hyperparamètre** fixé avant l'apprentissage.
* **Équation et définition des symboles :**
  $$\hat{y}_i = \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij} + \varepsilon_i$$
  * $\hat{y}_i$ : valeur prédite pour l'observation $i$ ;
  * $\beta_0$ : ordonnée à l'origine (biais) ;
  * $\beta_j$ : paramètre appris associé à la $j$-ème variable ;
  * $x_{ij}$ : valeur de la $j$-ème variable pour l'individu $i$ ;
  * $p$ : nombre total de variables explicatives ($p = 10$ ici) ;
  * $\varepsilon_i$ : terme d'erreur (résidu $y_i - \hat{y}_i$).

## 2. Ridge

**Variante C — question spécifique :** Comment évoluent les coefficients quand alpha augmente ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (régression linéaire régularisée $L_2$).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques standardisées $X \in \mathbb{R}^{n \times p}$ (*Diabetes*). En sortie, une prédiction continue $\hat{y} \in \mathbb{R}$.
* **Hypothèse clé :** Relation linéaire entre les entrées et la cible, avec une pénalisation sur la norme $L_2$ des coefficients pour réduire la variance en présence de multicolinéarité.
* **Étapes successives de calcul :** (1) Standardisation des variables explicatives (indispensable pour pénaliser équitablement chaque variable), (2) minimisation de la somme des carrés des résidus augmentée de la pénalité quadratique $\alpha \sum \beta_j^2$, (3) calcul des prédictions sur le jeu de test.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients rétrécis $\beta_j$ et l'intercept $\beta_0$. L'**hyperparamètre** est $\alpha > 0$ (`alpha`), fixé avant l'entraînement pour contrôler l'intensité de la régularisation.
* **Équation et définition des symboles :**
  $$\min_{\beta_0, \beta} \sum_{i=1}^{n} \left(y_i - \beta_0 - \sum_{j=1}^{p} \beta_j x_{ij}\right)^2 + \alpha \sum_{j=1}^{p} \beta_j^2$$
  * $y_i$ : valeur réelle de la cible pour l'individu $i$ ;
  * $\beta_0$ : ordonnée à l'origine (non pénalisée) ;
  * $\beta_j$ : coefficient appris de la variable $x_{ij}$ ;
  * $\alpha$ : hyperparamètre de régularisation $L_2$ ($\alpha \ge 0$) ;
  * $n, p$ : nombre d'observations et nombre de variables.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [données standardisées $X$, choix de $\alpha$] $\rightarrow$ Apprentissage [compromis entre erreur quadratique et contrainte circulaire $\sum \beta_j^2 \le t$] $\rightarrow$ Sortie [coefficients $\hat{\beta}$ réduits vers 0 sans s'annuler])*
* **Légende et choix algorithmique :** Ce schéma illustre la fonction objectif pénalisée $L_2$ qui stabilise les poids lorsque les variables sont corrélées.

### Contextes et choix (C3)
* **Contexte 1 (Finance) :** Modélisation du rendement d'un portefeuille ou prévision des ventes à partir de nombreux indicateurs macroéconomiques corrélés entre eux (inflation, taux d'intérêt, chômage).
* **Contexte 2 (Marketing) :** Mesure de l'impact de plusieurs canaux publicitaires (TV, radio, réseaux sociaux, presse) dont les budgets évoluent simultanément.
* **Structure de données attendue :** Données tabulaires numériques standardisées, particulièrement adaptées lorsque $p$ est grand ou que les variables sont fortement colinéaires.
* **Algorithme concurrent :** La régression **Lasso** (pénalité $L_1$), qui annule complètement certains coefficients au lieu de simplement les réduire.
* **Situation mal adaptée :** Lorsque seule une petite fraction des variables est réellement pertinente et qu'on souhaite une sélection automatique des variables (modèle parcimonieux).
* **Argumentation de la métrique :** La **MAE** permet de comparer directement l'erreur moyenne de prédiction avec la régression linéaire simple sur les mêmes données *Diabetes*.

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique :** Lorsque `alpha` augmente, la pénalité $L_2$ devient plus forte et **tous les coefficients $\beta_j$ se rapprochent progressivement de zéro (rétrécissement ou *shrinkage*)**, sans jamais devenir strictement nuls. Cela réduit la variance du modèle mais augmente son biais si `alpha` est trop élevé.
* **Effet de la modification d'hyperparamètre :** En testant `alpha = 0.1`, `alpha = 12.0` (référence) et `alpha = 500.0`, on observe qu'un très grand `alpha` écrase les coefficients (norme $L_2$ beaucoup plus faible) et fait remonter la MAE par sous-ajustement (*underfitting*), tandis qu'une régularisation modérée stabilise l'erreur par rapport à la régression linéaire sans pénalité.
* **Limite d'une seule exécution :** Le choix optimal de `alpha` ne peut pas être validé sur un seul découpage train/test ; il faut utiliser une validation croisée (`RidgeCV`) pour éviter tout biais lié à l'échantillonnage.

```python
# Test de sensibilité C4 : évolution des coefficients et de la MAE selon alpha
for a in [0.1, 12.0, 500.0]:
    mod = make_pipeline(StandardScaler(), Ridge(alpha=a)).fit(Xr_ent, yr_ent)
    norme_coef = np.linalg.norm(mod[-1].coef_)
    mae = mean_absolute_error(yr_test, mod.predict(Xr_test))
    print(f"alpha = {a:<6} -> MAE : {mae:.2f} | Norme L2 des coefficients : {norme_coef:.2f}")
```

## 3. Lasso

**Variante C — question spécifique :** Combien de variables sont annulées pour trois valeurs d’alpha ?

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (régression linéaire régularisée $L_1$).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques standardisées $X \in \mathbb{R}^{n \times p}$ (*Diabetes*). En sortie, une prédiction continue $\hat{y} \in \mathbb{R}$ accompagnée d'une sélection automatique de variables.
* **Hypothèse clé :** Relation linéaire entre les entrées et la cible, avec une hypothèse de parcimonie (*sparsity*) : seule une partie des variables explicatives influence réellement la cible.
* **Étapes successives de calcul :** (1) Standardisation des variables explicatives (obligatoire pour pénaliser chaque variable sur la même échelle), (2) minimisation de la somme des carrés des résidus augmentée de la pénalité $L_1$ via descente par coordonnées (*coordinate descent*), (3) mise à zéro stricte des coefficients les moins informatifs, (4) prédiction sur le jeu de test.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients $\beta_j$ (dont certains deviennent exactement nuls) et l'intercept $\beta_0$. L'**hyperparamètre** est $\alpha > 0$ (`alpha`), fixé avant l'entraînement pour contrôler le degré de parcimonie (ainsi que `max_iter` pour la convergence).
* **Équation et définition des symboles :**
  $$\min_{\beta_0, \beta} \frac{1}{2n} \sum_{i=1}^{n} \left(y_i - \beta_0 - \sum_{j=1}^{p} \beta_j x_{ij}\right)^2 + \alpha \sum_{j=1}^{p} |\beta_j|$$
  * $y_i$ : valeur réelle de la cible pour l'observation $i$ ;
  * $\beta_0$ : ordonnée à l'origine (non pénalisée) ;
  * $\beta_j$ : coefficient appris de la variable $x_{ij}$ ;
  * $\alpha$ : hyperparamètre de pénalisation $L_1$ ($\alpha \ge 0$) ;
  * $n, p$ : nombre d'observations et nombre de variables explicatives.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [données standardisées $X$, choix de $\alpha$] $\rightarrow$ Apprentissage [compromis entre erreur quadratique et contrainte en losange $\sum |\beta_j| \le t$ touchant les axes] $\rightarrow$ Sortie [vecteur creux $\hat{\beta}$ avec plusieurs coefficients exactement égaux à 0])*
* **Légende et choix algorithmique :** Ce schéma illustre la géométrie de la pénalité $L_1$ (losange dont les sommets sont sur les axes), expliquant pourquoi Lasso annule strictement certains coefficients contrairement à Ridge.

### Contextes et choix (C3)
* **Contexte 1 (Finance / Scoring) :** Sélection des ratios financiers les plus déterminants parmi des dizaines d'indicateurs comptables pour prédire le risque de défaut ou la rentabilité d'une entreprise.
* **Contexte 2 (Marketing) :** Identification des leviers d'acquisition clients réellement efficaces parmi un grand nombre de campagnes et de segments démographiques.
* **Structure de données attendue :** Données tabulaires numériques standardisées avec un grand nombre de variables potentiellement redondantes ou inutiles.
* **Algorithme concurrent :** **ElasticNet** (qui combine les pénalités $L_1$ et $L_2$) ou **Ridge**.
* **Situation mal adaptée :** Lorsque plusieurs variables très corrélées sont toutes importantes (Lasso a tendance à n'en garder arbitrairement qu'une seule et à annuler les autres).
* **Argumentation de la métrique :** La **MAE** permet d'évaluer si la suppression de certaines variables dégrade ou améliore la précision moyenne de prédiction par rapport à Ridge et à la régression linéaire.

  * Pour `alpha = 15.0` : **8 variables annulées** (seulement 2 coefficients non nuls conservés).
* **Effet de la modification d'hyperparamètre :** Plus `alpha` augmente, plus la contrainte $L_1$ est sévère et plus le modèle élimine de variables. Une régularisation modérée (`alpha = 2.0`) simplifie le modèle tout en gardant une bonne MAE (~41.5), tandis qu'un `alpha` trop élevé (`15.0`) provoque un sous-ajustement important (la MAE grimpe fortement) car trop d'information utile est supprimée.
* **Limite d'une seule exécution :** La sélection des variables retenues peut varier d'un échantillon à l'autre lorsque les variables sont corrélées ; une validation croisée (`LassoCV`) est requise pour stabiliser le choix de `alpha`.

```python
# Test de sensibilité C4 : nombre de variables annulées pour trois valeurs d'alpha
for a in [0.1, 2.0, 15.0]:
    mod = make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=15000)).fit(Xr_ent, yr_ent)
    non_nuls = np.count_nonzero(mod[-1].coef_)
    annules = len(mod[-1].coef_) - non_nuls
    mae = mean_absolute_error(yr_test, mod.predict(Xr_test))
    print(f"alpha = {a:<5} -> Variables annulées : {annules}/10 (conservées : {non_nuls}) | MAE : {mae:.2f}")
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (régression linéaire sur espace étendu par transformation polynomiale).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques $X \in \mathbb{R}^{n \times p}$ (*Diabetes*). En sortie, une prédiction continue $\hat{y} \in \mathbb{R}$ capturant des relations courbes et des interactions entre variables.
* **Hypothèse clé :** La variable cible dépend d'une combinaison linéaire des puissances et des produits croisés (interactions) des variables explicatives d'origine jusqu'à un degré $d$.
* **Étapes successives de calcul :** (1) Standardisation des variables initiales, (2) génération des nouveaux termes polynomiaux (carrés $x_j^2$ et interactions $x_j x_k$ pour $d=2$), (3) ajustement d'un modèle linéaire régularisé (ici Ridge) sur cette matrice étendue, (4) prédiction sur le jeu de test.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients $\beta$ associés à chaque terme initial, carré et produit croisé. Les **hyperparamètres** sont le **degré du polynôme** $d$ (`degree=2`) fixé avant la transformation, ainsi que la pénalité $\alpha$ de Ridge (`alpha=100.0`).
* **Équation et définition des symboles :**
  $$\hat{y}_i = \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij} + \sum_{j=1}^{p} \sum_{k=j}^{p} \beta_{jk} x_{ij} x_{ik} + \varepsilon_i \quad (\text{pour } d=2)$$
  * $\hat{y}_i$ : valeur prédite pour l'observation $i$ ;
  * $\beta_0$ : ordonnée à l'origine ;
  * $\beta_j$ : poids des effets linéaires simples ;
  * $\beta_{jk}$ : poids des termes quadratiques ($j=k$) et d'interaction ($j \neq k$) ;
  * $p$ : nombre initial de variables ($p=10$, générant 65 termes au degré 2) ;
  * $\varepsilon_i$ : terme d'erreur résiduelle.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [$X \in \mathbb{R}^{n \times 10}$, choix du degré $d$] $\rightarrow$ Apprentissage [expansion polynomiale $\Phi(X) \in \mathbb{R}^{n \times 65}$ puis régression Ridge] $\rightarrow$ Sortie [surface de régression courbe et prédictions $\hat{y}$])*
* **Légende et choix algorithmique :** Ce schéma montre que la non-linéarité est obtenue en projetant les données dans un espace de dimension supérieure avant d'y ajuster un hyperplan régularisé.

### Contextes et choix (C3)

* **Contexte 2 (Gestion de production / Logistique) :** Estimation du coût unitaire de fabrication en fonction du volume produit (courbe en U traduisant les économies d'échelle puis les coûts de saturation).
* **Structure de données attendue :** Tableau de variables numériques continues en nombre modéré (car le nombre de colonnes explose combinatoirement avec le degré $d$).
* **Algorithme concurrent :** Les **arbres de régression** ou les **forêts aléatoires**, qui capturent les non-linéarités sans explosion dimensionnelle.
* **Situation mal adaptée :** Lorsque le jeu de données contient beaucoup de variables ($p$ élevé) ou un degré $d \ge 3$, ce qui provoque une explosion combinatoire du nombre de colonnes et un fort surajustement.
* **Argumentation de la métrique :** Le coefficient de détermination **$R^2$** mesure la proportion de la variance totale de la cible expliquée par le modèle, ce qui permet de vérifier directement si l'ajout de termes courbes améliore l'explication des données par rapport au modèle linéaire ($R^2 = 0.485$).

### Test de sensibilité et interprétation (C4)

En revanche, sur le **jeu de test**, sans régularisation forte (`alpha` faible), le $R^2$ chute drastiquement à cause du surajustement ; avec une forte régularisation (`alpha=100.0`), le $R^2$ de test reste très proche du modèle linéaire simple (~0.49), prouvant que le gain se situe surtout sur l'entraînement.
* **Effet de la modification d'hyperparamètre :** En comparant `degree = 1`, `degree = 2` (référence) et `degree = 3` (avec `alpha=100.0`), on constate que `degree = 3` génère 285 variables : le $R^2$ d'entraînement grimpe encore, mais le $R^2$ de test diminue car le modèle mémorise le bruit de l'échantillon d'apprentissage.
* **Limite d'une seule exécution :** L'équilibre entre `degree` et `alpha` dépend fortement de la taille de l'échantillon d'entraînement ; une validation croisée est indispensable pour vérifier si les termes d'interaction apportent un réel gain généralisable.

```python
# Test de sensibilité C4 : comparaison entraînement vs test pour degree = 1, 2, 3
for d in [1, 2, 3]:
    mod = make_pipeline(StandardScaler(), PolynomialFeatures(degree=d, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
    r2_train = r2_score(yr_ent, mod.predict(Xr_ent))
    r2_test = r2_score(yr_test, mod.predict(Xr_test))
    nb_vars = mod[1].n_output_features_
    print(f"degree = {d} ({nb_vars:3d} variables) -> R² Entraînement : {r2_train:.3f} | R² Test : {r2_test:.3f}")
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (modèle non linéaire et non paramétrique par partitionnement récursif).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X \in \mathbb{R}^{n \times p}$ (*Diabetes*). En sortie, une prédiction continue par morceaux $\hat{y} \in \mathbb{R}$ correspondant à la moyenne des cibles dans chaque feuille terminale.
* **Hypothèse clé :** L'espace des variables peut être découpé séquentiellement en régions hyper-rectangulaires homogènes au sein desquelles la variable cible est approximée par une constante locale.
* **Étapes successives de calcul :** (1) À chaque nœud, recherche de la variable $j$ et du seuil $s$ qui minimisent la somme des erreurs quadratiques (MSE) des deux sous-groupes créés, (2) découpage binaire récursif de l'échantillon d'entraînement, (3) arrêt de la croissance selon les critères d'élagage (`max_depth`, `min_samples_leaf`), (4) attribution de la moyenne des observations de la feuille lors de la prédiction.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont la structure de l'arbre (variables choisies, seuils de coupure $s_m$ et moyennes prédites $c_m$ dans les feuilles). Les **hyperparamètres** sont les contraintes fixées avant l'apprentissage, comme la profondeur maximale (`max_depth=3`) et le nombre minimal d'observations par feuille (`min_samples_leaf=8`).
* **Équation et définition des symboles :**
  $$\hat{y}(x) = \sum_{m=1}^{M} c_m \mathbb{I}(x \in R_m) \quad \text{avec} \quad c_m = \frac{1}{N_m} \sum_{x_i \in R_m} y_i$$
  * $\hat{y}(x)$ : prédiction associée à l'observation $x$ ;
  * $M$ : nombre total de régions terminales (feuilles) de l'arbre ;
  * $R_m$ : $m$-ième région de l'espace définie par la suite des règles de coupure ;
  * $\mathbb{I}(x \in R_m)$ : fonction indicatrice valant $1$ si $x$ tombe dans la feuille $R_m$ et $0$ sinon ;
  * $c_m$ : moyenne des valeurs réelles $y_i$ des $N_m$ observations appartenant à $R_m$.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [racine contenant toutes les données $X, y$ et profondeur max fixée] $\rightarrow$ Apprentissage [coupures binaires successives $x_j \le s$ réduisant la variance interne] $\rightarrow$ Sortie [arbre final avec règles de décision et valeurs moyennes $c_m$ aux feuilles])*
* **Légende et choix algorithmique :** Ce schéma illustre le critère de séparation CART (minimisation de la variance intra-nœud) qui transforme l'espace continu en paliers constants.

### Contextes et choix (C3)
* **Contexte 1 (Commerce / Gestion des stocks) :** Estimation de la demande hebdomadaire d'un produit selon des règles métier lisibles (promotion active ou non, jour de la semaine, niveau de stock initial).
* **Contexte 2 (Finance / Immobilier) :** Évaluation rapide de la valeur d'un bien immobilier selon des seuils précis (surface $> 80\text{ m}^2$, quartier, présence d'ascenseur).
* **Structure de données attendue :** Tableau mixte (variables numériques ou catégorielles), robuste aux valeurs aberrantes en entrée et ne nécessitant aucune standardisation préalable.
* **Algorithme concurrent :** La **Forêt aléatoire en régression** (*Random Forest Regressor*), qui agrège plusieurs arbres pour réduire fortement l'instabilité et la variance.
* **Situation mal adaptée :** Lorsque la relation sous-jacente est lisse et continue (comme une tendance linéaire pure), car l'arbre approche une droite par une fonction en escalier grossière et ne sait pas extrapoler au-delà des valeurs vues à l'entraînement.
* **Argumentation de la métrique :** La **MAE** permet d'évaluer l'erreur moyenne absolue de ces prédictions par paliers et de la comparer directement aux régressions linéaires sur *Diabetes*.

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (effet de `max_depth` sur biais et variance) :** Une faible profondeur (`max_depth = 2`) limite le nombre de coupures : l'arbre est trop simple, ce qui génère un **biais élevé** (sous-ajustement sur l'entraînement et le test) mais une **faible variance**. À l'inverse, une grande profondeur (`max_depth = 12`) permet à l'arbre de mémoriser les cas particuliers : le **biais chute** sur l'entraînement (MAE très basse), mais la **variance explose** (surajustement), dégradant la MAE sur le jeu de test.
* **Effet de la modification d'hyperparamètre :** En testant `max_depth = 2`, `3` (référence) et `12` (avec `min_samples_leaf=2`), on observe que la MAE d'entraînement baisse continuellement avec la profondeur, alors que la MAE de test atteint son optimum autour d'une profondeur modérée (`max_depth = 3`) avant de remonter nettement à `max_depth = 12`.
* **Limite d'une seule exécution :** Un arbre de décision seul est très instable : une légère modification de l'échantillon d'entraînement peut changer la coupure racine et toute la structure de l'arbre ; une validation croisée est indispensable pour choisir `max_depth`.

```python
# Test de sensibilité C4 : impact de max_depth sur le compromis biais-variance
for d in [2, 3, 12]:
    mod = DecisionTreeRegressor(max_depth=d, min_samples_leaf=2, random_state=GRAINE).fit(Xr_ent, yr_ent)
    mae_train = mean_absolute_error(yr_ent, mod.predict(Xr_ent))
    mae_test = mean_absolute_error(yr_test, mod.predict(Xr_test))
    print(f"max_depth = {d:<2} -> MAE Entraînement : {mae_train:5.2f} | MAE Test : {mae_test:5.2f}")
```

## 6. Forêt aléatoire — régression

**Variante C — question spécifique :** La forêt réduit-elle la variance par rapport à l’arbre ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (méthode d'ensemble non linéaire par *bagging* d'arbres de décision).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X \in \mathbb{R}^{n \times p}$ (*Diabetes*). En sortie, une prédiction continue $\hat{y} \in \mathbb{R}$ obtenue par la moyenne des prédictions de l'ensemble des arbres.
* **Hypothèse clé :** L'agrégation (moyenne) de plusieurs arbres de régression profonds et décorrélés (entraînés sur des échantillons *bootstrap* différents et des sous-ensembles aléatoires de variables) réduit fortement la variance sans augmenter le biais.
* **Étapes successives de calcul :** (1) Tirage aléatoire avec remise (*bootstrap*) de $B$ sous-échantillons d'entraînement, (2) construction d'un arbre de régression sur chaque échantillon en sélectionnant à chaque nœud un sous-ensemble aléatoire de variables (`max_features`), (3) calcul de la prédiction de chaque arbre $T_b(x)$, (4) agrégation finale par moyenne arithmétique.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont l'ensemble des règles de coupure (variables, seuils) et des valeurs prédites dans les feuilles des $B$ arbres. Les **hyperparamètres** sont le nombre d'arbres (`n_estimators=60`), la profondeur maximale (`max_depth=6`), ou le nombre maximal de variables testées à chaque nœud (`max_features`).
* **Équation et définition des symboles :**
  $$\hat{y}_{\text{RF}}(x) = \frac{1}{B} \sum_{b=1}^{B} T_b(x)$$
  * $\hat{y}_{\text{RF}}(x)$ : prédiction finale de la forêt aléatoire pour l'observation $x$ ;
  * $B$ : nombre total d'arbres dans la forêt (`n_estimators`) ;
  * $T_b(x)$ : prédiction individuelle du $b$-ième arbre de régression entraîné sur le $b$-ième échantillon *bootstrap*.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [$X, y$, choix de $B$ arbres et `max_depth`] $\rightarrow$ Apprentissage [$B$ tirages bootstrap en parallèle + arbres entraînés sur sous-ensembles aléatoires de variables] $\rightarrow$ Sortie [agrégation par moyenne $\frac{1}{B}\sum T_b(x)$])*
* **Légende et choix algorithmique :** Ce schéma illustre le mécanisme de *bagging* (*Bootstrap Aggregating*) combiné à la sélection aléatoire de variables, permettant de décorréler les arbres et de lisser les erreurs individuelles.

### Contextes et choix (C3)
* **Contexte 1 (Finance / Assurance) :** Estimation du montant moyen des sinistres automobiles ou immobiliers à partir de dizaines de caractéristiques clients et géographiques.
* **Contexte 2 (Logistique / Supply Chain) :** Prévision de la demande quotidienne par référence produit en intégrant la météo, les jours fériés, les promotions et les historiques de vente.
* **Structure de données attendue :** Données tabulaires structurées (numériques ou catégorielles encodées), avec relations complexes non linéaires et interactions entre variables, sans besoin de standardisation.
* **Algorithme concurrent :** Le **Gradient Boosting en régression** (*HistGradientBoostingRegressor*), qui construit les arbres séquentiellement pour corriger les résidus précédents plutôt qu'en parallèle.
* **Situation mal adaptée :** Lorsque l'objectif principal est l'interprétabilité directe d'une formule simple ou l'extrapolation de tendances croissantes hors de la plage des valeurs d'entraînement.
* **Argumentation de la métrique :** La **MAE** permet de mesurer directement la baisse de l'erreur moyenne de prédiction par rapport à l'arbre de régression unique (algorithme n°5).

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (réduction de variance par rapport à l'arbre) :** Oui, la forêt aléatoire réduit nettement la variance par rapport à un arbre seul. En répétant l'entraînement sur 10 graines aléatoires différentes (code ci-dessous), l'écart-type de la MAE sur le jeu de test est beaucoup plus faible pour la forêt aléatoire que pour un arbre de décision unique, car la moyenne de 60 arbres lisse les fluctuations liées à l'échantillonnage.
* **Effet de la modification d'hyperparamètre :** En faisant varier `n_estimators` (`5`, `60` et `200` arbres), on observe qu'avec très peu d'arbres (`n_estimators=5`), la MAE est instable et plus élevée ; dès `n_estimators=60` (référence) ou `200`, l'erreur se stabilise et le gain de précision devient marginal au-delà d'un certain seuil, tout en augmentant le temps de calcul.
* **Limite d'une seule exécution :** Sur un seul découpage train/test, une petite forêt peut par hasard obtenir un bon score ; comparer l'écart-type sur plusieurs graines ou via validation croisée est indispensable pour prouver la réduction de variance.

```python
# Test de sensibilité C4 : stabilité (variance) arbre seul vs forêt + impact de n_estimators
maes_arbre, maes_foret = [], []
for seed in range(10):
    a = DecisionTreeRegressor(max_depth=6, random_state=seed).fit(Xr_ent, yr_ent)
    f = RandomForestRegressor(n_estimators=60, max_depth=6, n_jobs=-1, random_state=seed).fit(Xr_ent, yr_ent)
    maes_arbre.append(mean_absolute_error(yr_test, a.predict(Xr_test)))
    maes_foret.append(mean_absolute_error(yr_test, f.predict(Xr_test)))

print(f"Arbre seul (10 graines) -> MAE moy : {np.mean(maes_arbre):.2f} | Écart-type (variance) : {np.std(maes_arbre):.2f}")
print(f"Forêt (10 graines)      -> MAE moy : {np.mean(maes_foret):.2f} | Écart-type (variance) : {np.std(maes_foret):.2f}\n")

for n_est in [5, 60, 200]:
    mod = RandomForestRegressor(n_estimators=n_est, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
    print(f"n_estimators = {n_est:<3} -> MAE Test : {mean_absolute_error(yr_test, mod.predict(Xr_test)):.2f}")
```

## 7. Gradient Boosting — régression

**Variante C — question spécifique :** Que change max_iter sur le coût et le risque de surajustement ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (méthode d'ensemble séquentielle par *boosting* de gradient).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X \in \mathbb{R}^{n \times p}$ (*Diabetes*). En sortie, une prédiction continue $\hat{y} \in \mathbb{R}$ construite par addition progressive de petits arbres de régression.
* **Hypothèse clé :** Un ensemble d'apprenants faibles (arbres peu profonds) entraînés séquentiellement, où chaque nouvel arbre corrige les erreurs résiduelles (le gradient négatif de la fonction de perte) commises par la somme des arbres précédents.
* **Étapes successives de calcul :** (1) Initialisation du modèle par une constante (la moyenne $\bar{y}$ pour une perte quadratique), (2) à chaque itération $m$, calcul des pseudo-résidus (gradient négatif de la perte), (3) ajustement d'un petit arbre $h_m(x)$ sur ces résidus, (4) mise à jour de la prédiction globale pondérée par le taux d'apprentissage $\eta$.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont la prédiction initiale $F_0$ et l'ensemble des règles de coupure et valeurs aux feuilles de chaque arbre $h_m$. Les **hyperparamètres** sont le nombre maximal d'itérations/arbres (`max_iter=100`), le nombre maximal de feuilles par arbre (`max_leaf_nodes=9`) et le taux d'apprentissage (`learning_rate`).
* **Équation et définition des symboles :**
  $$F_m(x) = F_{m-1}(x) + \eta \, h_m(x) \quad \text{avec} \quad h_m \approx -\frac{\partial L(y, F_{m-1}(x))}{\partial F_{m-1}(x)}$$
  * $F_m(x)$ : prédiction cumulée du modèle à l'itération $m$ ;
  * $\eta$ : taux d'apprentissage (*learning rate*, $0 < \eta \le 1$) contrôlant la contribution de chaque arbre ;
  * $h_m(x)$ : $m$-ième arbre faible ajusté sur les résidus courants ;
  * $L(y, F)$ : fonction de perte différentiable (ex. : erreur quadratique $\frac{1}{2}(y - F)^2$).

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [$F_0(x) = \bar{y}$] $\rightarrow$ Apprentissage [boucle séquentielle : calcul des résidus $r_m \rightarrow$ arbre $h_m$ $\rightarrow$ mise à jour $F_m = F_{m-1} + \eta h_m$] $\rightarrow$ Sortie [prédiction finale $F_M(x)$])*
* **Légende et choix algorithmique :** Ce schéma illustre l'apprentissage séquentiel additif (descente de gradient dans l'espace des fonctions), contrairement à la forêt aléatoire qui entraîne ses arbres indépendamment en parallèle.

### Contextes et choix (C3)
* **Contexte 1 (Finance / Crédit) :** Prévision de la perte en cas de défaut (LGD) ou scoring de risque sur des données clients tabulaires complexes.
* **Contexte 2 (Commerce / Grande distribution) :** Prévision des ventes promotionnelles par magasin et catégorie de produits en intégrant de nombreuses variables calendaires et tarifaires.

En sortie, une probabilité d'appartenance à la classe positive $\hat{p}_i \in [0, 1]$ et l'étiquette binaire $\hat{y}_i \in \{0, 1\}$.
* **Hypothèse clé :** Une séquence de petits arbres de décision peu profonds (*weak learners*) construits successivement permet de corriger étape par étape les erreurs de probabilité (le gradient négatif de la *log-loss*) commises par l'ensemble précédent.
* **Étapes successives de calcul :** (1) Initialisation des scores bruts (*log-odds*) par une constante basée sur la proportion de la classe positive, (2) à chaque itération $m$, calcul des pseudo-résidus $r_{im} = y_i - \hat{p}_{i, m-1}$, (3) ajustement d'un arbre $h_m(x)$ sur ces résidus, (4) mise à jour additive du score $F_m(x) = F_{m-1}(x) + \eta h_m(x)$ et conversion en probabilité via la fonction sigmoïde.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont la constante initiale $F_0$ et l'ensemble des règles de coupure et valeurs aux feuilles de chaque arbre séquentiel $h_m$. * **Équation et définition des symboles :**
  $$F_m(x) = F_{m-1}(x) + \eta \, h_m(x) \quad \text{et} \quad \hat{p}_m(x) = \frac{1}{1 + \exp(-F_m(x))}$$
  * $F_m(x)$ : score cumulé (*log-odds*) à l'itération $m$ ;
  * $\eta$ : taux d'apprentissage (*learning rate*, $0 < \eta \le 1$) pondérant la contribution de chaque nouvel arbre ;
  * $h_m(x)$ : $m$-ième arbre ajusté sur le gradient négatif de l'entropie croisée binaire ;
  * $\hat{p}_m(x)$ : probabilité prédite après $m$ itérations.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (classification binaire linéaire à vaste marge).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques standardisées $X \in \mathbb{R}^{n \times p}$ (les 30 caractéristiques cellulaires de *Breast Cancer*). En sortie, une classe binaire $\hat{y}_i \in \{-1, +1\}$ (ou $\{0, 1\}$) déterminée par le signe de la fonction de décision par rapport à l'hyperplan séparateur.
* **Hypothèse clé :** Les deux classes sont séparables (ou quasi-séparables moyennant une pénalité sur les erreurs) par un hyperplan linéaire, et la frontière optimale est celle qui **maximise la marge** (la distance géométrique) avec les observations les plus proches appelées *vecteurs de support*.
* **Étapes successives de calcul :** (1) Standardisation obligatoire des variables explicatives pour que la distance euclidienne à l'hyperplan soit équitable sur toutes les dimensions, (2) résolution d'un problème d'optimisation quadratique convexe minimisant $\frac{1}{2}\|w\|^2 + C \sum \xi_i$ sous contrainte de marge, (3) identification des vecteurs de support définissant $w$ et $b$, (4) calcul du signe de $w^\top x_i + b$ sur le jeu de test.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont le vecteur normal à l'hyperplan $w \in \mathbb{R}^p$ et le biais $b \in \mathbb{R}$ (ainsi que la sélection des vecteurs de support). L'**hyperparamètre** est le coût de pénalisation des erreurs de marge $C > 0$ (`C`), fixé avant l'entraînement.
* **Équation et définition des symboles :**
  $$\min_{w, b, \xi} \frac{1}{2}\|w\|^2 + C \sum_{i=1}^{n} \xi_i \quad \text{sous contrainte} \quad y_i(w^\top x_i + b) \ge 1 - \xi_i, \;\; \xi_i \ge 0$$
  $$\hat{y}(x) = \text{signe}(w^\top x + b)$$
  * $w$ : vecteur normal à l'hyperplan séparateur (la largeur de la marge vaut $\frac{2}{\|w\|}$) ;
  * $b$ : ordonnée à l'origine (biais) ;
  * $\xi_i$ : variables ressorts (*slack variables*) mesurant le degré de violation de la marge par l'observation $i$ ;
  * $C$ : hyperparamètre contrôlant le compromis entre largeur de marge et tolérance aux erreurs de classement.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [$X$ standardisé, choix de $C$] $\rightarrow$ Apprentissage [maximisation de la marge $\frac{2}{\|w\|}$ entre les deux droites parallèles passant par les vecteurs de support] $\rightarrow$ Sortie [hyperplan central $w^\top x + b = 0$ et classement $\hat{y}$])*
* **Légende et choix algorithmique :** Ce schéma met en évidence la notion de « rue la plus large » (marge maximale) et le rôle exclusif des vecteurs de support situés sur ou à l'intérieur de la marge.

### Contextes et choix (C3)
* **Contexte 1 (Marketing / Analyse textuelle) :** Classification automatique de tickets du service client (réclamation urgente vs demande standard) ou analyse de sentiment sur des avis clients représentés en sacs de mots (TF-IDF en grande dimension).
* **Contexte 2 (Finance / Conformité) :** Détection de documents financiers conformes ou non conformes lors d'audits automatisés sur de grands volumes de variables textuelles et numériques.
* **Structure de données attendue :** Données tabulaires numériques denses ou creuses (*sparse*), obligatoirement standardisées, particulièrement performant lorsque le nombre de variables $p$ est grand.

* **Situation mal adaptée :** Lorsque les classes ne sont pas linéairement séparables (frontière fortement courbe ou circulaire) ou sur des bases contenant plusieurs millions d'observations (temps d'entraînement quadratique/cubique pour les solveurs classiques).
* **Argumentation de la métrique :** Le **score F1** permet d'évaluer l'équilibre entre précision et rappel sur *Breast Cancer* et de comparer directement l'hyperplan à vaste marge avec la régression logistique (n°8).

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (effet de `C` sur la largeur de la marge et les vecteurs de support) :** Lorsque **`C` est petit** (`C = 0.01`), la pénalité sur les erreurs $\xi_i$ est faible : le modèle privilégie une **marge très large** (norme $\|w\|$ petite) au prix d'un plus grand nombre de points situés dans la marge ou mal classés (**nombre élevé de vecteurs de support**). À l'inverse, lorsque **`C` est grand** (`C = 100.0`), toute erreur est sévèrement pénalisée : la **marge devient étroite** (norme $\|w\|$ grande), le **nombre de vecteurs de support diminue**, et l'hyperplan s'ajuste étroitement aux points frontières au risque de surajuster le bruit.
* **Effet de la modification d'hyperparamètre :** En testant `C = 0.01`, `C = 1.0` et `C = 100.0` sur *Breast Cancer*, on constate que le nombre de vecteurs de support chute nettement quand `C` augmente, tandis que le score F1 atteint son meilleur niveau pour une régularisation modérée (`C = 0.01` à `1.0`, F1 $\approx 0.978$–$0.986$) et baisse légèrement à `C = 100.0` par surajustement.

**Variante C — question spécifique :** Que change learning_rate quand n_estimators reste fixe ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (classification non linéaire à vaste marge par astuce du noyau / *kernel trick*).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques standardisées $X \in \mathbb{R}^{n \times p}$ (les 30 caractéristiques cellulaires de *Breast Cancer*). En sortie, une classe prédite $\hat{y}_i \in \{-1, +1\}$ (ou $\{0, 1\}$) issue d'une frontière de décision courbe et flexible dans l'espace d'origine.
* **Hypothèse clé :** Deux classes non séparables linéairement dans l'espace initial deviennent séparables par un hyperplan lorsqu'on les projette implicitement dans un espace de dimension supérieure (voire infinie) grâce à un noyau gaussien (*Radial Basis Function* — RBF) mesurant la similarité locale entre points.
* **Étapes successives de calcul :** (1) Standardisation obligatoire des variables d'entrée (le noyau RBF reposant sur la distance euclidienne $\|x - x_i\|^2$), (2) calcul de la matrice de similarité via le noyau gaussien $K(x_i, x_k)$, (3) résolution du problème dual d'optimisation quadratique pour trouver les multiplicateurs $\alpha_i$ des vecteurs de support, (4) prédiction par combinaison pondérée des similarités avec les vecteurs de support.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les multiplicateurs de Lagrange $\alpha_i > 0$ associés aux vecteurs de support et le biais $b \in \mathbb{R}$. Les **hyperparamètres** sont le paramètre de régularisation $C > 0$ (`C`) et le coefficient de portée du noyau $\gamma > 0$ (`gamma`), fixés avant l'entraînement.
* **Équation et définition des symboles :**
  $$K(x, x_i) = \exp\left(-\gamma \|x - x_i\|^2\right) \quad \text{et} \quad \hat{y}(x) = \text{signe}\left(\sum_{i \in \mathcal{S}} \alpha_i y_i K(x, x_i) + b\right)$$
  * $K(x, x_i)$ : noyau RBF gaussien mesurant la similarité entre $x$ et le vecteur de support $x_i$ ($K \in ]0, 1]$) ;
  * $\gamma$ : hyperparamètre contrôlant l'inverse du rayon d'influence d'un point d'entraînement ;

### Présentation (C1)
* **Catégorie :** Apprentissage non supervisé (partitionnement / *clustering* centroïdal par minimisation de l'inertie intra-classe).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques standardisées $X \in \mathbb{R}^{n \times p}$ (les 13 mesures physico-chimiques de *Wine*, sans étiquettes $y$). En sortie, une affectation de chaque observation à un groupe $\hat{c}_i \in \{0, \dots, K-1\}$ ainsi que les coordonnées des $K$ centroïdes $\mu_k \in \mathbb{R}^p$.
* **Hypothèse clé :** Les observations s'organisent naturellement en $K$ groupes sphériques (convexes) et de tailles comparables autour de centres de gravité appelés centroïdes, la proximité étant mesurée par la distance euclidienne.
* **Étapes successives de calcul :** (1) Standardisation indispensable des variables (pour qu'aucune unité ne domine la distance), (2) initialisation de $K$ centroïdes (méthode *k-means++*), (3) **étape d'affectation** : rattachement de chaque point $x_i$ au centroïde $\mu_k$ le plus proche au sens euclidien, (4) **étape de mise à jour** : recalcul de chaque centroïde comme la moyenne des points affectés à son cluster, (5) répétition des étapes 3 et 4 jusqu'à stabilisation des centroïdes.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coordonnées des $K$ centroïdes finaux $\mu_1, \dots, \mu_K \in \mathbb{R}^p$ (et l'inertie finale atteinte). L'**hyperparamètre** principal est le nombre de clusters $K$ (`n_clusters=3`), ainsi que le nombre d'initialisations aléatoires (`n_init`).
* **Équation et définition des symboles :**
  $$\min_{C_1, \dots, C_K} \sum_{k=1}^{K} \sum_{x_i \in C_k} \Vert{}x_i - \mu_k\Vert{}^2 \quad \text{avec} \quad \mu_k = \frac{1}{\vert{}C_k\vert{}} \sum_{x_i \in C_k} x_i$$
  * $K$ : nombre de clusters fixé a priori ;
  * $C_k$ : ensemble des observations affectées au $k$-ième cluster ;
  * $\mu_k$ : centroïde (vecteur moyenne) du cluster $C_k$ ;
  * $\Vert{}x_i - \mu_k\Vert{}^2$ : distance euclidienne au carré entre le point $x_i$ et son centroïde (la somme totale représente l'**inertie intra-cluster**).

*(Insérez ici votre schéma : Initialisation [$X$ standardisé, choix de $K$ et placement initial des centroïdes $\mu_k$] $\rightarrow$ Apprentissage [boucle itérative : affectation au centroïde le plus proche $\leftrightarrow$ recalcul des moyennes $\mu_k$] $\rightarrow$ Sortie [$K$ clusters disjoints et centroïdes finaux])*
* **Légende et choix algorithmique :** Ce schéma illustre l'algorithme itératif de Lloyd (alternance affectation / mise à jour des centres) qui converge vers un minimum local de l'inertie intra-classe.

* **Argumentation de la métrique :** Le **score de silhouette** (compris entre $-1$ et $+1$) mesure à la fois la cohésion interne d'un cluster et sa séparation avec le cluster voisin le plus proche, permettant d'évaluer la qualité de la partition sans recourir aux vraies étiquettes.

* **Hypothèse clé :** Les observations proches dans l'espace des caractéristiques peuvent être regroupées de manière ascendante (*bottom-up*), en fusionnant à chaque étape les deux groupes les plus similaires selon un critère d'agrégation (*linkage*).

### Présentation (C1)
* **Catégorie :** Apprentissage non supervisé (*clustering* par densité spatiale avec détection native du bruit).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques standardisées $X \in \mathbb{R}^{n \times p}$ (les 13 variables physico-chimiques de *Wine*). En sortie, une étiquette de cluster $\hat{c}_i \in \{0, 1, \dots, K-1\}$ pour les points denses, et l'étiquette $-1$ pour les points isolés considérés comme du bruit (*outliers*).
* **Hypothèse clé :** Un cluster est une région continue de haute densité de points dans l'espace des variables, séparée des autres clusters par des régions de faible densité, quelle que soit sa forme géométrique.
* **Étapes successives de calcul :** (1) Standardisation obligatoire des variables explicatives (la notion de voisinage reposant sur un rayon euclidien), (2) pour chaque point $x_i$, comptage des voisins situés dans une hypersphère de rayon $\varepsilon$, (3) classification des points en trois catégories : **points cœurs** (*core points*, ayant au moins `min_samples` voisins dans le rayon $\varepsilon$), **points frontières** (*border points*, dans le voisinage d'un point cœur sans en être un) et **points de bruit** ($-1$), (4) propagation par connexité des clusters à partir des points cœurs voisins.
* **Paramètre appris vs Hyperparamètre :** DBSCAN est un algorithme non paramétrique sans centroïdes : il **apprend** uniquement le partitionnement final (affectation aux clusters et identification du bruit) ainsi que le nombre automatique de clusters $K$. Les **hyperparamètres** sont le rayon du voisinage $\varepsilon > 0$ (`eps`) et le nombre minimal de points requis pour former une région dense (`min_samples`).
* **Équation et définition des symboles :**
  $$N_\varepsilon(x_i) = \{x_j \in X \mid d(x_i, x_j) \le \varepsilon\} \quad \text{et} \quad |N_\varepsilon(x_i)| \ge \text{min\_samples} \implies x_i \text{ est un point cœur}$$
  * $N_\varepsilon(x_i)$ : voisinage de rayon $\varepsilon$ autour de l'observation $x_i$ ;
  * $d(x_i, x_j)$ : distance (généralement euclidienne) entre $x_i$ et $x_j$ ;
  * $\varepsilon$ (`eps`) : rayon maximal d'un voisinage ;
  * $\text{min\_samples}$ : seuil minimal de densité (incluant le point $x_i$ lui-même).

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [$X$ standardisé, choix du rayon $\varepsilon$ et du seuil `min_samples`] $\rightarrow$ Apprentissage [cercles de rayon $\varepsilon$ autour de chaque point $\rightarrow$ identification des points cœurs, frontières et points isolés $-1$] $\rightarrow$ Sortie [clusters de formes arbitraires par fusion des voisinages denses + bruit $-1$])*
* **Légende et choix algorithmique :** Ce schéma illustre la propagation par densité (points cœurs reliés de proche en proche) et la mise à l'écart naturelle des points isolés sans forcer leur rattachement à un groupe.

### Contextes et choix (C3)
* **Contexte 1 (Marketing / Géomarketing) :** Identification de zones commerciales à forte concentration de clientèle ou de points de vente sur des données GPS urbaines (formes irrégulières suivant les axes routiers ou le littoral).
* **Contexte 2 (Finance / Détection d'anomalies) :** Repérage simultané des profils de transactions usuels (clusters denses) et des opérations suspectes atypiques (points étiquetés $-1$).
* **Structure de données attendue :** Données numériques denses et standardisées, de dimension faible à modérée, pouvant contenir du bruit et des clusters de formes non sphériques (allongées, courbes).
* **Algorithme concurrent :** **OPTICS** ou **HDBSCAN** (qui s'affranchissent d'un rayon $\varepsilon$ unique) ainsi que **K-Means** (pour des groupes sphériques sans bruit).
* **Situation mal adaptée :** En haute dimension ($p$ élevé, comme les 13 variables de *Wine* où les distances euclidiennes s'uniformisent par le « fléau de la dimension ») ou lorsque les clusters ont des densités très différentes les unes des autres.
* **Argumentation de la métrique :** Le **score de silhouette** (calculé en surveillant la proportion de points classés en bruit $-1$) permet d'évaluer la cohésion des groupes denses découverts sans supervision.

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (impact du rayon `eps` sur le nombre de clusters et le bruit) :** En dimension 13 (*Wine*), DBSCAN est extrêmement sensible au choix de **`eps`** : si **`eps` est trop petit** (`eps = 1.5`), aucun point n'a suffisamment de voisins dans ce rayon, et **100 % des observations sont classées comme du bruit** (`label = -1`, $0$ cluster trouvé). À l'inverse, si **`eps` est trop grand** (`eps = 4.0`), tous les voisinages se rejoignent et **DBSCAN fusionne la quasi-totalité des vins en un seul cluster unique** ($1$ cluster, $\approx 0\%$ de bruit). Ce n'est que dans une fenêtre étroite intermédiaire (`eps` autour de `2.2`–`2.5` avec `min_samples=5`) que DBSCAN parvient à dégager plusieurs clusters distincts tout en conservant une part modérée de bruit.
* **Effet de la modification d'hyperparamètre :** En faisant varier `eps` (`1.5`, `2.3`, `2.6` et `4.0` avec `min_samples=5`), on observe directement la transition brutale : tout en bruit (`eps=1.5`), émergence de 3 clusters (`eps=2.3`), fusion partielle (`eps=2.6`), puis un seul bloc global (`eps=4.0`).
* **Limite d'une seule exécution :** Sur des données en dimension $p \ge 10$, régler `eps` à l'aveugle est très délicat ; il faut tracer le graphique des distances au $k$-ième plus proche voisin (*k-distance graph*) ou réduire d'abord la dimension via ACP avant d'appliquer DBSCAN.

## 20. Mélanges gaussiens

**Variante C — question spécifique :** Comparez covariance_type = full et diag.

### Présentation (C1)
* **Catégorie :** Apprentissage non supervisé (modèle de mélange génératif probabiliste / *clustering* souple).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques standardisées $X \in \mathbb{R}^{n \times p}$ (les 13 variables physico-chimiques de *Wine*). En sortie, une matrice de responsabilités (probabilités d'appartenance à chacun des $K$ clusters) $\gamma_{ik} \in [0, 1]$ ainsi que l'étiquette la plus probable $\hat{c}_i = \arg\max_k \gamma_{ik}$.
* **Hypothèse clé :** Les données sont générées par un mélange de $K$ lois normales multivariées (gaussiennes), chacune caractérisée par son propre centre (moyenne) et sa propre forme ellipsoïdale (matrice de covariance).
* **Étapes successives de calcul :** (1) Standardisation des variables d'entrée, (2) initialisation des paramètres des $K$ gaussiennes (poids $\pi_k$, moyennes $\mu_k$, matrices de covariance $\Sigma_k$), (3) **étape E (Expectation)** : calcul de la probabilité a posteriori $\gamma_{ik}$ que chaque point $x_i$ provienne de la composante $k$, (4) **étape M (Maximization)** : mise à jour de $\pi_k, \mu_k, \Sigma_k$ en pondérant par $\gamma_{ik}$, (5) itération de l'algorithme EM jusqu'à convergence de la log-vraisemblance.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les poids de mélange $\pi_k$, les vecteurs moyennes $\mu_k \in \mathbb{R}^p$ et les matrices de covariance $\Sigma_k \in \mathbb{R}^{p \times p}$ de chaque composante. Les **hyperparamètres** sont le nombre de composantes (`n_components=3`), la structure des matrices de covariance (`covariance_type='full'`, `'tied'`, `'diag'`, `'spherical'`) et le nombre d'initialisations (`n_init`).
* **Équation et définition des symboles :**
  $$p(x_i) = \sum_{k=1}^{K} \pi_k \, \mathcal{N}(x_i \mid \mu_k, \Sigma_k) \quad \text{et} \quad \gamma_{ik} = \frac{\pi_k \, \mathcal{N}(x_i \mid \mu_k, \Sigma_k)}{\sum_{j=1}^{K} \pi_j \, \mathcal{N}(x_i \mid \mu_j, \Sigma_j)}$$
  * $\pi_k$ : poids a priori de la $k$-ième gaussienne ($\sum_{k=1}^K \pi_k = 1$) ;
  * $\mathcal{N}(x_i \mid \mu_k, \Sigma_k)$ : densité de probabilité gaussienne multivariée de moyenne $\mu_k$ et de covariance $\Sigma_k$ ;
  * $\gamma_{ik}$ : responsabilité (probabilité d'appartenance souple) du cluster $k$ pour l'observation $x_i$.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [$X$ standardisé, choix de $K$ gaussiennes et de `covariance_type`] $\rightarrow$ Apprentissage [algorithme EM : Étape E (calcul des probabilités $\gamma_{ik}$) $\leftrightarrow$ Étape M (mise à jour des ellipses $\mu_k, \Sigma_k$)] $\rightarrow$ Sortie [$K$ clusters elliptiques + probabilités d'appartenance])*
* **Légende et choix algorithmique :** Ce schéma illustre le *clustering* probabiliste souple (chaque point reçoit une probabilité d'appartenir à chaque ellipse gaussienne, contrairement à l'affectation rigide de K-Means).

### Contextes et choix (C3)
* **Contexte 1 (Marketing / Segmentation comportementale nuancée) :** Identification de profils clients hybrides (ex. : client à 70 % « acheteur régulier » et à 30 % « chasseur de promotions ») grâce aux probabilités d'appartenance souples.
* **Contexte 2 (Finance / Régimes de marché) :** Modélisation des régimes de volatilité des rendements financiers (régime calme vs régime de crise) par un mélange de distributions gaussiennes.
* **Structure de données attendue :** Données tabulaires numériques continues et standardisées, pouvant former des nuages de points allongés ou orientés obliquement (ellipses).
* **Algorithme concurrent :** **K-Means** (qui correspond au cas particulier limite d'un GMM avec matrices de covariance sphériques identiques et affectation dure).
* **Situation mal adaptée :** En très haute dimension avec peu d'observations lorsque `covariance_type='full'` (le nombre de paramètres de covariance explose en $\mathcal{O}(K p^2)$, causant du surajustement ou des matrices singulières).
* **Argumentation de la métrique :** Le **score de silhouette** évalue la séparation géométrique des clusters résultants, tandis que le **BIC** (*Bayesian Information Criterion*) permet de pénaliser la complexité des matrices de covariance.

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (effet de `covariance_type` et comparaison avec K-Means) :** Contrairement à K-Means qui impose des clusters sphériques de même rayon, le GMM adapte la forme des clusters selon **`covariance_type`** : `'spherical'` (sphères de rayon libre), `'diag'` (ellipses alignées sur les axes), `'tied'` (une même ellipse orientée partagée par tous les clusters) et `'full'` (chaque cluster apprend sa propre orientation et corrélation). Sur *Wine* ($n=178, p=13$), `'diag'` et `'spherical'` atteignent un score de silhouette très proche de K-Means ($\approx 0,285$), tandis que `'full'` estime $3 \times \frac{13 \times 14}{2} = 273$ paramètres de covariance sur seulement 178 lignes, ce qui le rend plus flexible mais plus sensible à l'initialisation (d'où un BIC plus élevé).
* **Effet de la modification d'hyperparamètre :** En comparant `'full'`, `'tied'`, `'diag'` et `'spherical'` avec `n_components=3`, on observe l'impact direct sur le critère **BIC** (plus il est bas, meilleur est le compromis vraisemblance / parcimonie) et le **score de silhouette**.
* **Limite d'une seule exécution :** L'algorithme EM pouvant converger vers un optimum local de vraisemblance selon les moyennes initiales, il faut impérativement fixer `n_init=10` et `random_state=GRAINE` pour stabiliser les résultats.

## 21. ACP

**Variante C — question spécifique :** Combien de composantes faut-il pour 80 % de variance ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


### Présentation (C1)
* **Catégorie :** Apprentissage non supervisé (réduction de dimension linéaire par projection orthogonale).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques centrées et réduites $X \in \mathbb{R}^{n \times p}$ (les 13 variables physico-chimiques de *Wine* dans `X_clu`). En sortie, une matrice de coordonnées projetées $Z \in \mathbb{R}^{n \times d}$ ($d < p$, par exemple $d = 2$ ou $3$ composantes principales orthogonales) maximisant la variance conservée.
* **Hypothèse clé :** L'information principale d'un jeu de données corrélées réside dans quelques directions orthogonales de variance maximale (axes principaux), tandis que les axes de faible variance correspondent essentiellement à du bruit ou de la redondance.
* **Étapes successives de calcul :** (1) Standardisation obligatoire des variables (moyenne nulle, écart-type unitaire), (2) calcul de la matrice de covariance (ou de corrélation) $\Sigma = \frac{1}{n} X^\top X$, (3) décomposition en valeurs propres et vecteurs propres (ou SVD) pour ordonner les axes par variance décroissante ($\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_p$), (4) projection des données sur les $d$ premiers vecteurs propres $Z = X W_d$.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont la matrice de projection $W_d \in \mathbb{R}^{p \times d}$ (les *loadings* ou vecteurs propres associés aux plus grandes valeurs propres) et la variance expliquée par chaque axe. L'**hyperparamètre** est le nombre de composantes conservées `n_components` (un entier $d$ ou une proportion de variance cible, ex. `0.90`).
* **Équation et définition des symboles :**
  $$Z = X W_d \quad \text{avec} \quad \Sigma w_k = \lambda_k w_k \quad \text{et} \quad \text{Part de variance} = \frac{\sum_{k=1}^{d} \lambda_k}{\sum_{j=1}^{p} \lambda_j}$$
  * $X$ : matrice des données standardisées ($n \times p$) ;
  * $W_d = [w_1, \dots, w_d]$ : matrice des $d$ premiers vecteurs propres unitaires et orthogonaux ;
  * $\lambda_k$ : $k$-ième valeur propre mesurant la variance projetée sur l'axe $k$ ;
  * $Z$ : nouvelles composantes principales décorrélées ($n \times d$).

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [$X$ standardisé de dimension $p=13$, choix de $d$ composantes] $\rightarrow$ Apprentissage [calcul des axes orthogonaux de variance maximale PC1, PC2 via décomposition de la covariance] $\rightarrow$ Sortie [projection des points dans le plan réduit $(PC1, PC2)$ et cumul de variance expliquée])*
* **Légende et choix algorithmique :** Ce schéma illustre la rotation orthogonale des axes vers les directions d'allongement maximal du nuage de points afin de compresser l'information en limitant la perte de variance.

### Contextes et choix (C3)
* **Contexte 1 (Finance / Gestion de portefeuille et taux) :** Extraction des facteurs macroéconomiques latents de la courbe des taux d'intérêt (niveau, pente, courbure) à partir de dizaines de maturités obligataires corrélées.
* **Contexte 2 (Marketing / Enquêtes de satisfaction) :** Synthèse d'un questionnaire client comportant 40 questions fortement corrélées en 3 ou 4 axes factoriels interprétables avant une segmentation K-Means.
* **Structure de données attendue :** Données tabulaires numériques continues, impérativement standardisées (sans standardisation, une variable exprimée en milliers dominerait artificiellement PC1).
* **Algorithme concurrent :** **t-SNE** ou **UMAP** (réductions non linéaires dédiées à la visualisation de voisinages locaux) ou **SVD tronquée** (pour les matrices creuses).
* **Situation mal adaptée :** Lorsque les structures intéressantes des données sont fortement non linéaires (ex. : spirales ou variétés courbes) ou lorsque l'on a besoin de conserver directement le sens métier brut de chaque variable individuelle.
* **Argumentation de la métrique :** La **variance expliquée cumulée** (et l'erreur quadratique de reconstruction) mesure objectivement la proportion d'information statistique conservée après projection en dimension réduite.

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (nombre de composantes pour atteindre 80 % et 90 % de variance sur *Wine*) :** Sur les 13 variables physico-chimiques standardisées de *Wine* (`X_clu`), les deux premières composantes expliquent à elles seules environ **55,4 %** de la variance totale (`PC1` $\approx 36,2\%$, `PC2` $\approx 19,2\%$). Il faut conserver **`5` composantes principales** pour franchir le seuil de **80 %** de variance expliquée ($\approx 80,2\%$), et **`8` composantes principales** pour dépasser **90 %** ($\approx 92,0\%$), permettant ainsi de supprimer près de 40 % des dimensions tout en gardant plus de 90 % de l'information.
* **Effet de la modification d'hyperparamètre (`n_components`) :** En faisant varier `n_components` de `2` à `10`, on observe la croissance décroissante de la variance expliquée cumulée (éboulis des valeurs propres) et la diminution simultanée de l'erreur moyenne de reconstruction $X - Z W_d^\top$.
* **Limite d'une seule exécution :** L'ACP étant un algorithme linéaire maximisant la variance globale, elle est sensible aux valeurs aberrantes extrêmes (qui tirent les premiers axes principaux) et ne garantit pas que les axes de plus forte variance soient toujours les plus discriminants pour une tâche supervisée ultérieure.

```python
# Test de sensibilité C4 : variance expliquée cumulée et seuils 80% / 90% sur X_clu
from sklearn.decomposition import PCA

pca_full = PCA(random_state=GRAINE).fit(X_clu)
var_cum = np.cumsum(pca_full.explained_variance_ratio_) * 100

for d in range(1, 11):
    X_proj = pca_full.transform(X_clu)[:, :d]
    X_rec = X_proj @ pca_full.components_[:d]
    err_rec = np.mean((X_clu - X_rec) ** 2)
    print(f"n_components = {d:<2} -> Variance cumulée : {var_cum[d-1]:5.1f}% | Erreur reconstruction (MSE) : {err_rec:.3f}")

nb_80 = np.argmax(var_cum >= 80) + 1
nb_90 = np.argmax(var_cum >= 90) + 1
print(f"\nComposantes requises -> >= 80% : {nb_80} | >= 90% : {nb_90}")
```

## 22. t-SNE

**Variante C — question spécifique :** La structure change-t-elle quand perplexity passe de 12 à 30 ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


### Présentation (C1)
* **Catégorie :** Apprentissage non supervisé (réduction de dimension linéaire par décomposition en valeurs singulières tronquée — *Latent Semantic Analysis* / LSA).
* **Nature des entrées / sorties :** En entrée, une matrice de données $X \in \mathbb{R}^{n \times p}$ (dense standardisée comme `X_clu` ou creuse / *sparse* sans centrage obligatoire). En sortie, une matrice de représentation réduite $Z_d = U_d \Sigma_d \in \mathbb{R}^{n \times d}$ ($d < p$) conservant les $d$ plus grandes valeurs singulières.
* **Hypothèse clé :** Une matrice de données de grande dimension peut être approximée fidèlement par une factorisation de rang faible $d$ ne retenant que les $d$ directions singulières associées aux plus grandes valeurs singulières.
* **Étapes successives de calcul :** (1) Mise à l'échelle des variables (si données denses), (2) décomposition de la matrice $X$ sans la centrer explicitement (préservant la structure creuse des zéros le cas échéant) en $X \approx U_d \Sigma_d V_d^\top$, (3) sélection des $d$ plus grandes valeurs singulières $\sigma_1 \ge \dots \ge \sigma_d$, (4) projection des observations par $Z_d = X V_d = U_d \Sigma_d$.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les vecteurs singuliers droits $V_d^\top \in \mathbb{R}^{d \times p}$ (les axes latents) et les valeurs singulières $\sigma_1, \dots, \sigma_d$. Les **hyperparamètres** sont le nombre de composantes conservées `n_components` ($d$), l'algorithme de calcul (`algorithm='randomized'` ou `'arpack'`) et le nombre d'itérations (`n_iter`).
* **Équation et définition des symboles :**
  $$X \approx X_d = U_d \Sigma_d V_d^\top \quad \text{et} \quad Z_d = X V_d = U_d \Sigma_d$$
  * $U_d \in \mathbb{R}^{n \times d}$ : matrice des $d$ vecteurs singuliers gauches orthonormés ;
  * $\Sigma_d = \text{diag}(\sigma_1, \dots, \sigma_d)$ : matrice diagonale des $d$ plus grandes valeurs singulières ;
  * $V_d \in \mathbb{R}^{p \times d}$ : matrice des $d$ vecteurs singuliers droits (directions de projection) ;

### Schéma original (C2)

### Présentation (C1)
* **Catégorie :** Apprentissage non supervisé (détection d'anomalies par ensemble d'arbres d'isolation aléatoires — *iTrees*).
* **Nature des entrées / sorties :** En entrée, une matrice de variables numériques $X \in \mathbb{R}^{n \times p}$ (les 13 variables physico-chimiques de *Wine* dans `X_clu`). En sortie, un score d'anomalie continu $s(x, n) \in [0, 1]$ (ou `decision_function`) ainsi qu'une étiquette binaire $\hat{y}_i \in \{+1, -1\}$ où $+1$ désigne une observation normale et $-1$ une anomalie.
* **Hypothèse clé :** Les anomalies sont à la fois **rares** (*few*) et **très différentes** (*different*) des observations normales : lors d'un partitionnement aléatoire récursif de l'espace, elles se retrouvent isolées beaucoup plus rapidement (avec un chemin moyen plus court depuis la racine de l'arbre) que les points normaux regroupés dans des zones denses.
* **Étapes successives de calcul :** (1) Tirage de $T$ sous-échantillons aléatoires de taille $\psi$ (`max_samples`), (2) construction de $T$ arbres d'isolation (*iTrees*) en choisissant à chaque nœud une variable $j$ au hasard et un seuil de coupure uniforme entre $\min(x_j)$ et $\max(x_j)$ jusqu'à isoler chaque point, (3) calcul de la longueur moyenne du chemin $E(h(x))$ sur l'ensemble des $T$ arbres, (4) normalisation pour obtenir le score d'anomalie $s(x, \psi)$ et seuillage selon le taux de `contamination`.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont la structure des $T$ arbres aléatoires (les variables et seuils tirés à chaque nœud, la longueur des chemins) et le seuil de décision déduit du taux de contamination. * **Équation et définition des symboles :**
  $$s(x, \psi) = 2^{-\frac{E(h(x))}{c(\psi)}} \quad \text{avec} \quad c(\psi) = 2H(\psi - 1) - \frac{2(\psi - 1)}{\psi}$$
  * $h(x)$ : longueur du chemin (nombre d'arêtes traversées de la racine à la feuille isolant $x$) ;
  * $E(h(x))$ : longueur moyenne du chemin de $x$ sur les $T$ arbres d'isolation ;
  * $c(\psi)$ : longueur moyenne théorique dans un arbre binaire de recherche de taille $\psi$ ($H(i) \approx \ln(i) + 0,5772$ étant le nombre harmonique) ;
  * $s(x, \psi) \in [0, 1]$ : score d'anomalie ($s \to 1$ si $E(h(x)) \to 0$, soit une anomalie franche ; $s \approx 0,5$ pour un point normal).

  * $\text{lrd}_k(x_i)$ : densité locale d'atteignabilité (*Local Reachability Density*) autour de $x_i$ ;
  * $\text{LOF}_k(x_i) \approx 1$ : $x_i$ a la même densité que ses voisins (point normal) ;

* **Hypothèse clé :** La série évolue autour d'un niveau local qui varie lentement au cours du temps, sans tendance structurelle ni cycle saisonnier, et les observations les plus récentes portent davantage d'information pertinente pour l'avenir que les observations anciennes.

* **Équation et définition des symboles :**

  * $\alpha \in ]0, 1[$ : paramètre de lissage contrôlant la décroissance exponentielle des poids $(1-\alpha)^k$ accordés au passé ;
  * $\hat{y}_{T+h|T}$ : prévision à l'horizon $h$ calculée à la date $T$.

### Schéma original (C2)

* **Argumentation de la métrique :** Le **RMSE** (ou le **MAE**) sur un découpage chronologique (*train* sur le passé, *test* sur les dernières périodes) mesure l'écart moyen de prévision dans l'unité d'origine de la série sans fuite temporelle.

À l'inverse, lorsque **`alpha` est élevé** (`alpha = 0.90`), le modèle suit presque point par point la dernière observation ($\hat{y}_{t+1} \approx y_t$, proche d'une marche aléatoire naïve), capturant le bruit aléatoire. **Limite fondamentale :** comme la prévision future est constante ($\hat{y}_{T+h|T} = \ell_T$), sur une série comportant une **tendance** (même légère), le lissage exponentiel simple sous-estime systématiquement les valeurs futures croissantes.

* **Limite d'une seule exécution :** Sur une seule fenêtre de test courte, un `alpha` élevé peut sembler performant par hasard si les dernières valeurs de test restent proches de $y_T$ ; une validation croisée glissante (*rolling-origin / TimeSeriesSplit*) est recommandée.

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


### Présentation (C1)
* **Catégorie :** Analyse de séries temporelles (modélisation stochastique linéaire autorégressive intégrée à moyenne mobile avec composante saisonnière).
* **Nature des entrées / sorties :** En entrée, une série chronologique univariée ordonnée dans le temps $(y_1, y_2, \dots, y_T) \in \mathbb{R}^T$ (et éventuellement des variables exogènes $X_t$). En sortie, une prévision ponctuelle $\hat{y}_{T+h|T}$ accompagnée de ses intervalles de confiance probabilistes.
* **Hypothèse clé :** Après différenciation ordinaire (ordre $d$) et saisonnière (ordre $D$ de période $s$) pour rendre la série **stationnaire** (moyenne et autocovariance constantes dans le temps), la valeur courante s'explique linéairement par ses propres valeurs passées (partie AR) et par les chocs aléatoires passés (partie MA).
* **Étapes successives de calcul :** (1) Diagnostic de stationnarité (test de Dickey-Fuller augmenté — ADF) et application des opérateurs de différenciation $\Delta^d$ et $\Delta_s^D$, (2) identification des ordres $(p, d, q) \times (P, D, Q, s)$ via les corrélogrammes ACF / PACF ou la minimisation du critère AIC, (3) estimation des coefficients par maximum de vraisemblance (filtre de Kalman), (4) vérification que les résidus se comportent comme un bruit blanc non corrélé, (5) prévision récursive à l'horizon $h$.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients autorégressifs $(\phi_1, \dots, \phi_p, \Phi_1, \dots, \Phi_P)$, les coefficients moyenne mobile $(\theta_1, \dots, \theta_q, \Theta_1, \dots, \Theta_Q)$, la constante $c$ et la variance du bruit blanc $\sigma^2$. Les **hyperparamètres** sont les ordres non saisonniers `order=(p, d, q)` et saisonniers `seasonal_order=(P, D, Q, s)`.
* **Équation et définition des symboles :**
  $$\Phi_P(L^s)\,\phi_p(L)\,(1 - L^s)^D (1 - L)^d y_t = c + \Theta_Q(L^s)\,\theta_q(L)\,\varepsilon_t, \quad \varepsilon_t \sim \mathcal{N}(0, \sigma^2)$$
  * $L$ : opérateur retard (*lag operator*) tel que $L y_t = y_{t-1}$ et $L^s y_t = y_{t-s}$ ;
  * $(1 - L)^d$ et $(1 - L^s)^D$ : différenciations ordinaire d'ordre $d$ et saisonnière d'ordre $D$ de période $s$ ;
  * $\phi_p(L), \Phi_P(L^s)$ : polynômes autorégressifs (AR) ordinaire d'ordre $p$ et saisonnier d'ordre $P$ ;
  * $\theta_q(L), \Theta_Q(L^s)$ : polynômes moyenne mobile (MA) ordinaire d'ordre $q$ et saisonnier d'ordre $Q$ ;
  * $\varepsilon_t$ : bruit blanc gaussien (innovation aléatoire à l'instant $t$).

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [série $y_t$, choix des ordres $(p,d,q)\times(P,D,Q,s)$] $\rightarrow$ Apprentissage [stationnarisation par différenciation $\Delta^d \Delta_s^D \rightarrow$ estimation des poids AR ($\phi$) et MA ($\theta$) sur les retards $\rightarrow$ diagnostic de bruit blanc des résidus] $\rightarrow$ Sortie [prévision $\hat{y}_{T+h|T}$ et intervalle de confiance à 95 %])*
* **Légende et choix algorithmique :** Ce schéma résume la méthodologie de Box-Jenkins : stationnariser la série par différenciation, modéliser la dépendance temporelle (AR + MA), puis valider l'absence d'autocorrélation résiduelle.

### Contextes et choix (C3)
* **Contexte 1 (Macroéconomie / Finance) :** Prévision de l'inflation mensuelle, des ventes au détail nationales ou des revenus trimestriels d'une entreprise cotée en tenant compte des cycles annuels ($s = 4$ ou $12$).
* **Contexte 2 (Énergie / Logistique) :** Prévision de la demande énergétique mensuelle ou des volumes de fret en intégrant à la fois l'inertie récente (AR/MA) et les effets calendaires saisonniers (et éventuellement des variables exogènes météo avec SARIMAX).
* **Structure de données attendue :** Série temporelle univariée continue et régulièrement espacée, suffisamment longue (idéalement $T \ge 50$ à $100$ points et plusieurs cycles saisonniers) avec une structure linéaire après différenciation.
* **Algorithme concurrent :** Le **Lissage de Holt-Winters** (plus rapide et intuitif sur des séries régulières) ou le **Gradient Boosting / Régression sur retards (*lags*)** (plus flexible pour capturer des non-linéarités complexes et de nombreuses variables exogènes).
* **Situation mal adaptée :** Sur des séries à très haute fréquence avec saisonnalités multiples simultanées (données horaires présentant des cycles journalier, hebdomadaire et annuel), sur des séries très courtes ($T < 30$), ou lorsque la dynamique est fortement non linéaire avec ruptures brutales.
* **Argumentation de la métrique :** L'**AIC** (*Akaike Information Criterion*) permet de sélectionner sur l'échantillon d'entraînement la spécification $(p,d,q)\times(P,D,Q,s)$ la plus parcimonieuse, tandis que le **RMSE** sur le jeu de test futur valide la précision réelle des prévisions hors échantillon.

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (rôle de la différenciation $d, D$ et comparaison des ordres $(p,d,q)$) :** L'ordre de différenciation **`d`** (et **`D`**) élimine les tendances et dérives saisonnières pour rendre la série stationnaire : un **`d = 0`** sur une série tendancielle produit une prévision qui retombe systématiquement vers la moyenne historique (forte sous-estimation), tandis qu'une **sur-différenciation** (`d = 2` inutilement) augmente artificiellement la variance des résidus et élargit les intervalles de confiance. De même, omettre la composante saisonnière (`seasonal_order=(0,0,0,0)`) laisse toute la vague cyclique dans les résidus, alors qu'un modèle **SARIMA** complet `(1,1,1)x(1,1,1,12)` abaisse fortement l'AIC et le RMSE de test.
* **Effet de la modification d'hyperparamètre :** En comparant quatre spécifications sur une série saisonnière avec tendance (`ARIMA(1,0,1)` sans différenciation, `ARIMA(1,1,1)` sans saisonnalité, `SARIMA(1,1,1)x(0,1,1,12)` et `SARIMA(2,1,2)x(1,1,1,12)`), on vérifie directement le gain massif apporté par la différenciation saisonnière $D=1$ et le compromis AIC / RMSE.
* **Limite d'une seule exécution :** L'estimation par maximum de vraisemblance peut rencontrer des problèmes de non-convergence numérique si les ordres $(p, q, P, Q)$ sont trop élevés sur un historique court ; il est préférable de privilégier des ordres parcimonieux ($p, q \le 2$) validés par l'AIC et le test de Ljung-Box sur les résidus.

```python
# Test de sensibilité C4 : rôle de la différenciation (d, D) et des ordres SARIMA
import warnings
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Série locale reproductible avec tendance + saisonnalité mensuelle (s=12) sur 6 ans (72 mois)
rng_ar = np.random.default_rng(GRAINE)
t_ar = np.arange(72)
serie_ar = pd.Series(100 + 1.2 * t_ar + 12 * np.sin(2 * np.pi * t_ar / 12) + rng_ar.normal(0, 2.5, 72))
ar_ent, ar_test = serie_ar.iloc[:60], serie_ar.iloc[60:]

configs_sarima = [
    ("ARIMA(1,0,1) [d=0, sans saison]", (1, 0, 1), (0, 0, 0, 0)),
    ("ARIMA(1,1,1) [d=1, sans saison]", (1, 1, 1), (0, 0, 0, 0)),
    ("SARIMA(1,1,1)x(0,1,1,12)", (1, 1, 1), (0, 1, 1, 12)),
    ("SARIMA(2,1,1)x(1,1,1,12)", (2, 1, 1), (1, 1, 1, 12)),
]

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    for nom_s, ord_ns, ord_s in configs_sarima:
        mod_sar = SARIMAX(ar_ent, order=ord_ns, seasonal_order=ord_s).fit(disp=False)
        prev_sar = mod_sar.forecast(len(ar_test))
        rmse_sar = np.sqrt(mean_squared_error(ar_test, prev_sar))
        print(f"{nom_s:<32} -> AIC : {mod_sar.aic:6.1f} | RMSE Test : {rmse_sar:.3f}")
```

```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

```python
# Comparaison de l'effet du taux d'apprentissage alpha sur SARSA
alphas = [0.01, 0.15, 0.5]
resultats_sarsa = {}

for a_val in alphas:
    q_table_tmp = np.zeros((env.observation_space.n, env.action_space.n))
    recompenses_episodes = []

    for episode in range(1100):
        etat, _ = env.reset(seed=GRAINE + episode)
        epsilon = max(.08, .8 * (1 - episode / 900))
        action = choisir_action(etat, epsilon)
        total_recompense = 0

        for pas in range(120):
            suivant, recompense, termine, tronque, _ = env.step(action)
            action_suivante = choisir_action(suivant, epsilon)
            cible = recompense + (0 if termine else .95 * q_table_tmp[suivant, action_suivante])
            q_table_tmp[etat, action] += a_val * (cible - q_table_tmp[etat, action])
            total_recompense += recompense
            etat, action = suivant, action_suivante
            if termine or tronque: break

        recompenses_episodes.append(total_recompense)

    resultats_sarsa[a_val] = {
        'q_moyenne': q_table_tmp.mean(),
        'derniers_episodes_moy': np.mean(recompenses_episodes[-100:])
    }
    print(f"Alpha = {a_val:<5} -> Valeur Q moyenne : {q_table_tmp.mean():.2f} | Récompense moy (100 derniers épisodes) : {np.mean(recompenses_episodes[-100:]):.2f}")
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé appliqué aux séries temporelles (reformulation d'une série chronologique en problème de régression tabulaire par fenêtrage glissant / *sliding window*).
* **Nature des entrées / sorties :** En entrée, une matrice de caractéristiques construite à partir des $p$ valeurs passées (retards ou *lags* $y_{t-1}, y_{t-2}, \dots, y_{t-p}$), éventuellement enrichie de statistiques glissantes (moyenne mobile) ou variables calendaires, $X_t \in \mathbb{R}^{n' \times p}$. En sortie, la valeur future à prévoir $\hat{y}_t \in \mathbb{R}$ (à un pas ou par prévision récursive multi-pas).
* **Hypothèse clé :** La dynamique future d'une série temporelle peut être apprise par n'importe quel algorithme de régression supervisée (Ridge, Random Forest, Gradient Boosting) dès lors que l'historique récent est encodé sous forme de variables explicatives retardées (*lags*).
* **Étapes successives de calcul :** (1) Transformation de la série univariée $(y_1, \dots, y_T)$ en matrice supervisée $(X_t, y_t)$ via un décalage temporel de $p$ pas (`shift(1)`, ..., `shift(p)`) et suppression des $p$ premières lignes incomplètes, (2) découpage **strictement chronologique** (entraînement sur le passé, test sur le futur, sans jamais mélanger aléatoirement les lignes), (3) entraînement du régresseur supervisé sur $(X_{\text{train}}, y_{\text{train}})$, (4) prévision sur le test (directe à un pas ou récursive pas à pas en réinjectant chaque prédiction $\hat{y}_{t+1}$ comme nouveau retard pour prévoir $\hat{y}_{t+2}$).
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont ceux du régresseur supervisé choisi (ex. : les coefficients $\beta_0, \beta_1, \dots, \beta_p$ pour une régression Ridge, ou les règles de coupure des arbres pour un Random Forest / Gradient Boosting). Les **hyperparamètres** sont le nombre de retards `n_lags` ($p$, taille de la fenêtre historique), les variables calendaires ou agrégats glissants ajoutés, ainsi que les hyperparamètres propres au régresseur (ex. : `alpha` pour Ridge, `n_estimators` et `max_depth` pour Random Forest).
* **Équation et définition des symboles :**
  $$\hat{y}_t = f_\theta\left(y_{t-1}, y_{t-2}, \dots, y_{t-p}, \, z_t\right) \quad \text{(ex. linéaire : } \hat{y}_t = \beta_0 + \sum_{k=1}^{p} \beta_k y_{t-k} + \gamma^\top z_t\text{)}$$
  * $y_{t-k}$ : valeur retardée (*lag*) de la série à l'instant $t - k$ ($k \in \{1, \dots, p\}$) ;
  * $p$ (`n_lags`) : profondeur de la fenêtre temporelle passée fournie au modèle ;
  * $z_t$ : variables exogènes ou calendaires optionnelles à l'instant $t$ (mois, jour de la semaine, promotion) ;
  * $f_\theta(\cdot)$ : fonction de régression supervisée apprise sur l'historique d'entraînement.

### Schéma original (C2)
*(Insérez ici votre schéma : Initialisation [série $y_t$, choix de la fenêtre de $p$ retards] $\rightarrow$ Apprentissage [construction de la table $(y_{t-1}, \dots, y_{t-p}) \to y_t$, découpage chronologique train/test, entraînement du modèle supervisé] $\rightarrow$ Sortie [prévisions futures $\hat{y}_{T+1}, \dots, \hat{y}_{T+h}$ par boucle récursive ou directe])*
* **Légende et choix algorithmique :** Ce schéma illustre le passage d'une série chronologique univariée à un tableau supervisé classique, permettant d'exploiter toute la puissance des algorithmes de Machine Learning tabulaire.

### Contextes et choix (C3)
* **Contexte 1 (E-commerce / Prévision des ventes multi-facteurs) :** Prévision de la demande quotidienne par magasin en combinant les ventes des 7 et 14 derniers jours (*lags*) avec des variables externes (prix, promotions en cours, jours fériés, météo).
* **Contexte 2 (Finance / Prévision de volatilité et de flux) :** Modélisation des volumes de transactions journaliers ou des indicateurs de risque en intégrant à la fois les retards récents de la série et des indicateurs macroéconomiques exogènes.
* **Structure de données attendue :** Série temporelle régulière suffisamment longue pour supporter la perte des $p$ premières lignes due au fenêtrage et conserver un échantillon d'entraînement riche.
* **Algorithme concurrent :** **SARIMAX** (modèle statistique linéaire natif pour séries temporelles) ou **Prophet**.
* **Situation mal adaptée :** Sur des séries très courtes ($T < 30$ points, où créer 12 ou 24 retards consomme la majorité des données), en prévision récursive à très long horizon sur des modèles à base d'arbres (les arbres ne peuvent pas extrapoler au-delà des valeurs maximales vues dans l'entraînement sans différenciation préalable), ou en cas de fuite de données (*data leakage*) si l'on utilise un `train_test_split` aléatoire.
* **Argumentation de la métrique :** Le **RMSE** sur la période de test future mesure l'erreur de prévision hors échantillon tout en garantissant le respect strict de la chronologie temporelle.

### Test de sensibilité et interprétation (C4)
* **Réponse à la question spécifique (effet du nombre de retards `n_lags`, choix du régresseur et prévision à un pas vs récursive) :** Le choix de **`n_lags`** ($p$) détermine l'horizon de mémoire accessible au modèle : sur une série présentant un cycle saisonnier de période $12$, un modèle avec **`n_lags = 2`** (mémoire trop courte) ignore le cycle annuel et obtient un RMSE de test médiocre, tandis que passer à **`n_lags = 12`** capture directement le retard saisonnier $y_{t-12}$ et divise l'erreur par plus de deux. Par ailleurs, la **prévision récursive multi-pas** (où chaque prédiction $\hat{y}_{T+k}$ est réinjectée comme entrée pour prédire $\hat{y}_{T+k+1}$) accumule les erreurs à mesure que l'horizon s'allonge et révèle une limite clé des forêts aléatoires : **un Random Forest ne sachant pas extrapoler au-delà du maximum observé dans l'entraînement**, il plafonne sur une tendance croissante non différenciée, contrairement à une **régression Ridge** qui extrapole linéairement la tendance.
* **Effet de la modification d'hyperparamètre :** En comparant `n_lags = 2`, `6` et `12` avec une **régression Ridge** puis avec un **Random Forest Regressor** en prévision récursive sur 12 mois futurs, on observe le gain massif de `n_lags = 12` et la capacité d'extrapolation supérieure de Ridge face au plafonnement de Random Forest sur une tendance croissante.
* **Limite d'une seule exécution :** La prévision récursive multi-pas étant sensible au cumul d'erreurs sur les premiers pas futurs, il est indispensable d'évaluer le modèle sur plusieurs origines temporelles glissantes (`TimeSeriesSplit`) et de stationnariser la tendance si l'on utilise des modèles à base d'arbres.

```python
# Test de sensibilité C4 : impact de n_lags (2, 6, 12) et comparaison Ridge vs Random Forest en prévision récursive
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor

# Série locale reproductible avec tendance + saisonnalité mensuelle (période 12) sur 7 ans (84 mois)
rng_lag = np.random.default_rng(GRAINE)
t_lag = np.arange(84)
serie_lag = 100 + 1.2 * t_lag + 12 * np.sin(2 * np.pi * t_lag / 12) + rng_lag.normal(0, 2.5, 84)
horizon = 12
y_train_s, y_test_s = serie_lag[:-horizon], serie_lag[-horizon:]

def creer_lags(serie, p):
    X_l, y_l = [], []
    for i in range(p, len(serie)):
        X_l.append(serie[i - p:i])
        y_l.append(serie[i])
    return np.array(X_l), np.array(y_l)

def prev_recursive(modele, historique, p, h):
    fenetre = list(historique[-p:])
    preds = []
    for _ in range(h):
        x_in = np.array(fenetre[-p:]).reshape(1, -1)
        y_hat = modele.predict(x_in)[0]
        preds.append(y_hat)
        fenetre.append(y_hat)
    return np.array(preds)

for nom_mod, reg in [
    ("Ridge (alpha=1.0)", Ridge(alpha=1.0)),
    ("RandomForest", RandomForestRegressor(n_estimators=100, random_state=GRAINE))
]:
    for p_lags in [2, 6, 12]:
        X_tr, y_tr = creer_lags(y_train_s, p_lags)
        reg.fit(X_tr, y_tr)
        preds_rec = prev_recursive(reg, y_train_s, p_lags, horizon)
        rmse_rec = np.sqrt(mean_squared_error(y_test_s, preds_rec))
        print(f"{nom_mod:<18} | n_lags = {p_lags:<2} -> RMSE Test récursif (12 pas) : {rmse_rec:.3f}")
```
